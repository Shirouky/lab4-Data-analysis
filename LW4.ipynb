# Базовые функции

```python
import pandas as pd

def import_xlsx():
  FILE_NAME = "DataFrame beads (3).xlsx"
  df = pd.read_excel(FILE_NAME)
  return df
```

```python
def save_xlsx(df):
  df.to_excel("DataFrame.xlsx", index=False)
```

```python
def print_distribution(df, column):
  counts = df[column].value_counts()
  print(f"Распределение {column} по категориям:")
  for category, count in counts.items():
      print(f"{category}: {count} объектов")
```

```python
df = import_xlsx()
```

# Задание 1

```python
def categorize_column(df, column):
  category_dict = {}
  print_distribution(df, column)
  unique_values = list(df[column].unique())

  for i in range(len(unique_values)):
    category_dict[unique_values[i]] = i + 1
  df[f"{column}_category"] = df[column].map(category_dict)

  print_distribution(df, f"{column}_category")
  return df
```

```python
df = categorize_column(df, "size")
```

# Задание 2

```python
def sampling_column(df, column, labels):
  print_distribution(df, column)
  df[f"{column}_category"] = pd.cut(df[column], bins=len(labels), labels=labels, include_lowest=True)
  print_distribution(df, f"{column}_category")
  print("\nПолучившиеся интервалы:")
  print(pd.cut(df[column], bins=len(labels), include_lowest=True).cat.categories)
  return df

df = sampling_column(df, "price", ["Низкая", "Средняя", "Высокая", "Очень высокая"])
```

```python
def get_binary(column):
  return pd.get_dummies(column.astype(str))

binary_df = get_binary(df["price_category"])
binary_df
```

```python
# combined_df = pd.concat([df, binary_df], axis=1)
combined_df = df.join(binary_df)
combined_df
```
